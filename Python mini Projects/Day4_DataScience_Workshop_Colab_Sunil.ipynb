# Introduction to NumPy

NumPy stands for **Numerical Python**.

It is mainly used for:
- Arrays
- Mathematical calculations
- Scientific computing
- Data analysis

NumPy arrays are:
- Faster than Python lists
- Memory efficient
- Easy to use


```

# Creating NumPy Arrays

import numpy as np

# 1D Array
arr1 = np.array([1, 2, 3, 4, 5])

print("1D Array:")
print(arr1)

# 2D Array
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\n2D Array:")
print(arr2)
```

```

# Array Properties

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# Shape -> Rows and Columns
print("Shape:", arr.shape)

# Number of dimensions
print("Dimensions:", arr.ndim)

# Total number of elements
print("Size:", arr.size)

# Data type
print("Data Type:", arr.dtype)
```


# NumPy Array Operations

## Types of Operations
- Arithmetic Operations
- Broadcasting
- Aggregation
- Comparison Operations


```

# Arithmetic Operations

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```

# Broadcasting Example

arr = np.array([1, 2, 3])

# Adding a single value to all elements
result = arr + 10

print(result)
```

```

# Aggregation Functions

numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
print("Standard Deviation:", np.std(numbers))
```

```

# Comparison Operations

arr = np.array([5, 10, 15, 20])

# Checking values greater than 10
print(arr > 10)

# Filtering values
print(arr[arr > 10])
```

# Introduction to Pandas

Pandas is a powerful library used for:
- Data analysis
- Data cleaning
- Working with tables

It was created by **Wes McKinney** in 2008.

## Advantages
- Reads CSV, Excel, JSON files easily
- Handles missing values
- Filters and sorts data
- Performs grouping and merging



# Pandas Series

A Series is:
- One-dimensional
- Similar to a column in Excel
- Has labels called indexes


```

# Creating a Pandas Series

data = pd.Series([10, 20, 30, 40])

print(data)
```

```

# Series with custom indexes

students = pd.Series(
    [85, 90, 78],
    index=["Ali", "Sara", "John"]
)

print(students)
```


# Pandas DataFrame

A DataFrame is:
- A table-like structure
- Made of rows and columns
- Similar to an Excel sheet


```

# Creating a DataFrame

data = {
    "Name": ["Ali", "Sara", "John", "Emma"],
    "Age": [21, 22, 20, 23],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

print(df)
```

```

# Basic DataFrame Operations

# Display first rows
print(df.head())

# Display information about dataset
print("\nDataset Info:")
print(df.info())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```

# Selecting Columns

print(df["Name"])

# Selecting multiple columns
print(df[["Name", "Marks"]])
```

```

# Filtering Data

# Students with marks greater than 80
high_marks = df[df["Marks"] > 80]

print(high_marks)
```

```

# Sorting Data

sorted_df = df.sort_values(by="Marks", ascending=False)

print(sorted_df)
```

```

# Handling Missing Values

data = {
    "Name": ["Ali", "Sara", None, "Emma"],
    "Marks": [85, None, 78, 88]
}

missing_df = pd.DataFrame(data)

print("Original Data:")
print(missing_df)

# Checking missing values
print("\nMissing Values:")
print(missing_df.isnull())

# Filling missing values
filled_df = missing_df.fillna("Not Available")

print("\nAfter Filling Missing Values:")
print(filled_df)
```


# Reading & Writing Data with Pandas


# Introduction to Matplotlib

Matplotlib is used to create:
- Line Charts
- Bar Charts
- Pie Charts
- Scatter Plots

It is one of the most popular visualization libraries in Python.


```

# Line Plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.figure(figsize=(6,4))

plt.plot(x, y, marker='o')

plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

```

# Bar Chart

students = ["Ali", "Sara", "John"]
marks = [85, 92, 78]

plt.figure(figsize=(6,4))

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```

# Pie Chart

sizes = [40, 35, 25]
labels = ["Python", "Java", "C++"]

plt.figure(figsize=(6,6))

plt.pie(sizes, labels=labels, autopct='%1.1f%%')

plt.title("Programming Language Popularity")

plt.show()
```

```

# Scatter Plot

x = [1, 2, 3, 4, 5]
y = [5, 7, 8, 10, 12]

plt.figure(figsize=(6,4))

plt.scatter(x, y)

plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```


# Introduction to Seaborn

Seaborn is built on top of Matplotlib.

It helps create:
- Beautiful statistical plots
- Advanced visualizations
- Heatmaps
- Distribution plots


```

# Loading sample dataset from Seaborn

tips = sns.load_dataset("tips")

print(tips.head())
```

```

# Seaborn Scatter Plot

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Total Bill vs Tip")

plt.show()
```

```

# Seaborn Bar Plot

plt.figure(figsize=(7,5))

sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Average Total Bill Per Day")

plt.show()
```

```

# Seaborn Histogram

plt.figure(figsize=(7,5))

sns.histplot(tips["total_bill"], bins=10)

plt.title("Distribution of Total Bill")

plt.show()
```


# Correlation & Heatmap

Correlation shows the relationship between columns.

- Positive Correlation → Both increase together
- Negative Correlation → One increases, other decreases


```

# Correlation Matrix

correlation = tips.corr(numeric_only=True)

print(correlation)
```

```

# Heatmap Visualization

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```


# Complete Data Analysis Workflow

## Steps:
1. Load Data
2. Explore Data
3. Clean Data
4. Analyze Data
5. Visualize Data
6. Draw Insights


```

# Mini Data Analysis Project

# Step 1: Load dataset
df = sns.load_dataset("tips")

# Step 2: Explore dataset
print(df.head())

# Step 3: Check missing values
print(df.isnull().sum())

# Step 4: Average bill amount
print("\nAverage Bill:", df["total_bill"].mean())

# Step 5: Visualization
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    x="day",
    y="total_bill"
)

plt.title("Bill Distribution by Day")

plt.show()
```
