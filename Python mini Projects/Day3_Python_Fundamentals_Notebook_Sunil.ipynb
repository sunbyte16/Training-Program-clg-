## Introduction to Python

```python
# Simple Python Program
print("Hello World!")
print("Welcome to Python Programming")
```

## Variables & Data Types

```python
# Variables store data

name = "Adnan"      # String
age = 22            # Integer
height = 5.9        # Float
is_student = True   # Boolean

print(name)
print(age)
print(height)
print(is_student)
```

## Checking Data Types

```python
# type() checks datatype

x = 10
y = 3.14
z = "Python"

print(type(x))
print(type(y))
print(type(z))
```

## Type Conversion

```python
# Type Conversion Example

num = "100"

converted_num = int(num)

print(converted_num)
print(type(converted_num))
```

## Operators in Python

```python
# Arithmetic Operators

a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
```

## Conditional Statements

```python
# if-else Example

number = 7

if number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

## Loops in Python

```python
# for loop

for i in range(1, 6):
    print(i)

print("----------------")

# while loop

count = 1

while count <= 5:
    print(count)
    count += 1
```

## Functions in Python

```python
# Function Example

def greet():
    print("Good Morning Students!")

greet()

# Function with Parameters

def add(a, b):
    print("Sum:", a + b)

add(10, 20)
```

## Types of Function Arguments

```python
# Keyword Arguments

def student(name, age):
    print(name, age)

student(age=21, name="Aman")

# Default Arguments

def country(name="India"):
    print(name)

country()
country("USA")

# Variable Length Arguments

def total(*numbers):
    print(sum(numbers))

total(10, 20, 30)
```

## Data Structures & Operations

```python
# LIST OPERATIONS

students = ["Aman", "Rahul", "Sara"]

students.append("John")      # Add item
students.remove("Rahul")     # Remove item
students.insert(1, "Priya")  # Insert at specific position
students.sort()              # Sort list
students.reverse()           # Reverse list

print("List:", students)
print("First Student:", students[0])
print("Total Students:", len(students))

print("----------------")

# TUPLE OPERATIONS

colors = ("Red", "Green", "Blue")

print("Tuple:", colors)
print("First Color:", colors[0])
print("Length of Tuple:", len(colors))
print("Count of Red:", colors.count("Red"))

print("----------------")

# SET OPERATIONS

numbers = {1, 2, 3, 3, 4}

numbers.add(5)
numbers.remove(2)
numbers.update([6, 7])   # Add multiple items
numbers.discard(10)      # No error if item not found

print("Set:", numbers)
print("Is 3 present?", 3 in numbers)
print("Total Elements:", len(numbers))

print("----------------")

# DICTIONARY OPERATIONS

student = {
    "name": "Aman",
    "age": 21
}

student["course"] = "Python"
student["city"] = "Delhi"

print("Dictionary:", student)

print("Student Name:", student["name"])
print("Student Age:", student.get("age"))

# Print all keys and values
print("Keys:", student.keys())
print("Values:", student.values())

# Remove an item
student.pop("city")

print("Updated Dictionary:", student)
```

## File Handling

```python
# Writing into file

with open("sample.txt", "w") as file:
    file.write("Hello Students")

# Reading file

with open("sample.txt", "r") as file:
    data = file.read()
    print(data)
```

## Error Handling

```python
# Exception Handling Example

try:
    a = 10
    b = 0

    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program Ended")
```

## Modules & Libraries

```python
# Using Math Module

import math

print(math.sqrt(25))
print(math.factorial(5))

print("----------------")

# Using NumPy Library

import numpy as np

arr = np.array([1, 2, 3, 4])

print(arr)
```

## Mini Project - Simple Calculator

```python
# Simple Calculator

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

operation = input("Enter operation (+, -, *, /): ")

if operation == "+":
    print("Result:", num1 + num2)

elif operation == "-":
    print("Result:", num1 - num2)

elif operation == "*":
    print("Result:", num1 * num2)

elif operation == "/":
    print("Result:", num1 / num2)

else:
    print("Invalid Operation")
```
